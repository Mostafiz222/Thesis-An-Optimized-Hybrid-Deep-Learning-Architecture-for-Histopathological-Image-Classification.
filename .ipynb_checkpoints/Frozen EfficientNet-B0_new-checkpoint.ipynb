```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
import torchvision.transforms as transforms
import torchvision.models as models
from medmnist import PathMNIST, INFO
import numpy as np
import time
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)
```

```python
info = INFO["pathmnist"]
num_classes = len(info["label"])

print("Dataset:", info["description"])
print("Number of classes:", num_classes)
print("Class labels:", info["label"])
```

```python
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor()
])

train_ds = PathMNIST(split="train", transform=transform, download=True)
val_ds   = PathMNIST(split="val",   transform=transform, download=True)
test_ds  = PathMNIST(split="test",  transform=transform, download=True)

print("Train samples:", len(train_ds))
print("Val samples:  ", len(val_ds))
print("Test samples: ", len(test_ds))
```

```python
train_loader = DataLoader(train_ds, batch_size=64, shuffle=True)
val_loader   = DataLoader(val_ds, batch_size=64, shuffle=False)
test_loader  = DataLoader(test_ds, batch_size=64, shuffle=False)
```

```python
effnet = models.efficientnet_b0(pretrained=True)

for param in effnet.parameters():
    param.requires_grad = False
```

```python
in_features = effnet.classifier[1].in_features
effnet.classifier = nn.Identity()

effnet = effnet.to(device)
effnet.eval()

print("Frozen EfficientNet-B0 loaded")
print("Embedding dim:", in_features)
```

```python
linear_probe = nn.Linear(in_features, num_classes).to(device)
print(linear_probe)
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(linear_probe.parameters(), lr=1e-3)
epochs = 3
```

```python
def run_epoch(backbone, head, loader, train=False):
    backbone.eval()
    head.train() if train else head.eval()

    total_loss, correct, total = 0, 0, 0

    with torch.set_grad_enabled(train):
        for x, y in loader:
            x = x.to(device)
            y = y.squeeze().long().to(device)

            feats = backbone(x)
            outputs = head(feats)
            loss = criterion(outputs, y)

            if train:
                optimizer.zero_grad()
                loss.backward()
                optimizer.step()

            total_loss += loss.item() * y.size(0)
            correct += (outputs.argmax(1) == y).sum().item()
            total += y.size(0)

    return total_loss / total, 100 * correct / total
```

```python
history = {
    "train_acc": [],
    "val_acc": [],
    "train_loss": [],
    "val_loss": []
}

start_time = time.time()

for epoch in range(1, epochs + 1):
    tr_loss, tr_acc = run_epoch(effnet, linear_probe, train_loader, train=True)
    va_loss, va_acc = run_epoch(effnet, linear_probe, val_loader)

    history["train_loss"].append(tr_loss)
    history["train_acc"].append(tr_acc)
    history["val_loss"].append(va_loss)
    history["val_acc"].append(va_acc)

    print(
        f"Epoch [{epoch:02d}/{epochs}] | "
        f"Train Acc: {tr_acc:.2f}% | Val Acc: {va_acc:.2f}%"
    )

print("Training time:", time.time() - start_time)
```

```python
test_loss, test_acc = run_epoch(effnet, linear_probe, test_loader)

print("\n===== FINAL TEST RESULT =====")
print(f"Test Loss:     {test_loss:.4f}")
print(f"Test Accuracy: {test_acc:.2f}%")
```

```python
effnet.eval()
linear_probe.eval()

all_preds = []
all_labels = []

with torch.no_grad():
    for x, y in test_loader:
        x = x.to(device)
        y = y.squeeze().long()

        feats = effnet(x)
        outputs = linear_probe(feats)
        preds = outputs.argmax(dim=1).cpu()

        all_preds.append(preds)
        all_labels.append(y)

y_pred = torch.cat(all_preds).numpy()
y_true = torch.cat(all_labels).numpy()

print("Total test samples:", len(y_true))
```

```python
classwise_acc = {}

for cls in range(num_classes):
    mask = (y_true == cls)
    correct = (y_pred[mask] == cls).sum()
    total = mask.sum()

    acc = correct / total if total > 0 else 0
    classwise_acc[cls] = acc

for cls, acc in classwise_acc.items():
    print(f"Class {cls} accuracy: {acc:.4f}")
```

```python
classes = list(classwise_acc.keys())
accuracies = [classwise_acc[c] * 100 for c in classes]

plt.figure(figsize=(8,5))
plt.bar(classes, accuracies)
plt.xlabel("Class label")
plt.ylabel("Accuracy (%)")
plt.title("Class-wise Accuracy – Frozen EfficientNet-B0 (Test Set)")
plt.ylim(0, 100)

for i, acc in enumerate(accuracies):
    plt.text(i, acc + 1, f"{acc:.1f}%", ha="center")

plt.tight_layout()
plt.show()
```

```python
cm = confusion_matrix(y_true, y_pred)
print("Confusion Matrix:\n", cm)
```

```python
cm_norm = cm.astype(float) / cm.sum(axis=1, keepdims=True)

plt.figure(figsize=(8,6))
sns.heatmap(
    cm_norm,
    annot=True,
    fmt=".2f",
    cmap="Reds",
    xticklabels=range(num_classes),
    yticklabels=range(num_classes)
)

plt.xlabel("Predicted label")
plt.ylabel("True label")
plt.title("Normalized Confusion Matrix – Frozen EfficientNet-B0")
plt.tight_layout()
plt.show()
```

```python
torch.save(
    {
        "model": "Frozen EfficientNet-B0 + Linear Probe",
        "dataset": "PathMNIST",
        "test_accuracy": test_acc
    },
    "efficientnetb0_test_accuracy.pth"
)

print("✔ Test accuracy saved")
```

```python
torch.save(
    {
        "classwise_accuracy": classwise_acc,
        "confusion_matrix_raw": cm,
        "confusion_matrix_normalized": cm_norm
    },
    "efficientnetb0_test_metrics.pth"
)

print("✔ Test metrics saved")
```
